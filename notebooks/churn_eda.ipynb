# Customer churn EDA

This notebook uses the same generator as `train.py`, then checks churn balance and feature relationships.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(str(Path.cwd().parent))
from train import make_churn_dataset

sns.set_theme(style="whitegrid")
df = make_churn_dataset(rows=6500, random_state=17)
df.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df, x="churned", ax=axes[0])
sns.boxplot(data=df, x="churned", y="tenure_months", ax=axes[1])
axes[0].set_title("Churn class balance")
axes[1].set_title("Tenure by churn outcome")
plt.tight_layout()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.barplot(data=df, x="contract_type", y="churned", ax=axes[0])
sns.barplot(data=df, x="payment_method", y="churned", ax=axes[1])
sns.histplot(data=df, x="monthly_charges", hue="churned", kde=True, ax=axes[2])
for ax in axes[:2]:
    ax.tick_params(axis="x", rotation=25)
plt.tight_layout()
```

```python
numeric = ["tenure_months", "monthly_charges", "support_tickets", "last_login_days", "churned"]
sns.heatmap(df[numeric].corr(), annot=True, cmap="vlag", center=0)
plt.title("Numeric correlation map")
plt.tight_layout()
```
